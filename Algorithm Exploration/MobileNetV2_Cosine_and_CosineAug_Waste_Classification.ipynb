# MobileNetV2 — Cosine & Cosine+Augment

Paper: Yong, L., Ma, L., Sun, D., & Du, L. (2023). *Application of MobileNetV2 to waste classification.*
PLOS ONE 18(3): e0282336. <https://doi.org/10.1371/journal.pone.0282336>

Notebook ini menjalankan **MobileNetV2** dengan **dua varian** dalam 1 file, pada dataset, split, dan
protokol training yang sama dengan eksperimen lain (split dari `outputs/splits/all.csv`, 8:1:1):

| Varian | Prefix artefak | Resep |
|---|---|---|
| B · cosine | `mobilenetv2_cosine_*` | cosine annealing, label smoothing 0.1, weight decay 1e-4, augmentasi standar |
| Cosine + Augment | `mobilenetv2_cosine_aug_*` | cosine annealing, label smoothing 0.1, weight decay 1e-4, augmentasi kuat (TrivialAugmentWide + ColorJitter + RandomErasing) |

File ini **baru** dan tidak menimpa hasil baseline MobileNetV2 (`mobilenetv2_trash30_*` / `mobilenetv2_*`) yang sudah ada.

## 3. Transform dan loader

```python
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

# Standar (untuk varian cosine)
train_transform = transforms.Compose([
    transforms.RandomResizedCrop(IMG_SIZE),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# Augmentasi kuat (untuk varian cosine+augment)
aug_train_transform = transforms.Compose([
    transforms.RandomResizedCrop(IMG_SIZE),
    transforms.RandomHorizontalFlip(),
    transforms.TrivialAugmentWide(),
    transforms.ColorJitter(0.2, 0.2, 0.2, 0.05),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
    transforms.RandomErasing(p=0.15, scale=(0.02, 0.15)),
])

eval_transform = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(IMG_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])


class TrashDataset(Dataset):
    def __init__(self, frame: pd.DataFrame, root: Path, transform=None):
        self.frame = frame.reset_index(drop=True)
        self.root = Path(root)
        self.transform = transform

    def __len__(self) -> int:
        return len(self.frame)

    def __getitem__(self, index: int):
        row = self.frame.iloc[index]
        with Image.open(self.root / row["rel_path"]) as img:
            image = img.convert("RGB")
        if self.transform is not None:
            image = self.transform(image)
        return image, int(row["label"])


def seed_worker(worker_id: int) -> None:
    worker_seed = torch.initial_seed() % (2 ** 32)
    np.random.seed(worker_seed)
    random.seed(worker_seed)


def make_loader(frame: pd.DataFrame, training: bool, batch_size: int, transform=None) -> DataLoader:
    dataset = TrashDataset(
        frame, DATA_ROOT,
        transform=transform if transform is not None else (train_transform if training else eval_transform),
    )
    drop_last = training and len(dataset) >= batch_size and len(dataset) % batch_size == 1
    generator = torch.Generator()
    generator.manual_seed(SEED)
    return DataLoader(
        dataset, batch_size=batch_size, shuffle=training,
        num_workers=NUM_WORKERS, pin_memory=torch.cuda.is_available(),
        drop_last=drop_last, persistent_workers=NUM_WORKERS > 0,
        worker_init_fn=seed_worker, generator=generator,
    )


train_df = df[df["split"] == "train"].reset_index(drop=True)
val_df = df[df["split"] == "val"].reset_index(drop=True)
test_df = df[df["split"] == "test"].reset_index(drop=True)

val_loader = make_loader(val_df, training=False, batch_size=BATCH_SIZE)
test_loader = make_loader(test_df, training=False, batch_size=TEST_BATCH_SIZE)

print(f"Train: {len(train_df):,} | Val: {len(val_df):,} | Test: {len(test_df):,}")
```

## 4. Model MobileNetV2

```python
def count_parameters(model: nn.Module) -> int:
    return sum(p.numel() for p in model.parameters() if p.requires_grad)


def build_mobilenet_v2(num_classes: int, pretrained: bool = True) -> nn.Module:
    weights = "IMAGENET1K_V1" if pretrained else None
    model = models.mobilenet_v2(weights=weights)
    in_features = model.classifier[1].in_features
    model.classifier[1] = nn.Linear(in_features, num_classes)
    return model


print("MobileNetV2 builder siap.")
```

## 5. Training loop

```python
def train_model(
    model, train_loader, val_loader, criterion, optimizer, scheduler,
    epochs, device, checkpoint_path, history_path, model_name, arch,
    extra_config=None, full_model_path=None,
    early_stopping_patience=None, early_stop_min_epochs=0,
):
    since = time.time()
    best_model_wts = copy.deepcopy(model.state_dict())
    best_val_acc = -1.0
    best_epoch = 0
    epochs_without_improvement = 0
    stopped_epoch = 0
    history = []
    dataloaders = {"train": train_loader, "val": val_loader}
    dataset_sizes = {phase: len(loader.dataset) for phase, loader in dataloaders.items()}

    for epoch in range(1, epochs + 1):
        print(f"Epoch {epoch}/{epochs}")
        print("-" * 10)
        epoch_start = time.time()
        epoch_losses, epoch_accs = {}, {}
        improved_this_epoch = False

        for phase in ["train", "val"]:
            model.train() if phase == "train" else model.eval()
            running_loss = 0.0
            running_corrects = 0
            loader = dataloaders[phase]
            progress = tqdm(total=dataset_sizes[phase], desc=f"{model_name} {phase} {epoch}/{epochs}", leave=False)
            for inputs, labels in loader:
                inputs = inputs.to(device, non_blocking=True)
                labels = labels.to(device, non_blocking=True)
                optimizer.zero_grad()
                with torch.set_grad_enabled(phase == "train"):
                    outputs = model(inputs)
                    _, preds = torch.max(outputs, 1)
                    loss = criterion(outputs, labels)
                    if phase == "train":
                        loss.backward()
                        optimizer.step()
                running_loss += loss.item() * inputs.size(0)
                running_corrects += torch.sum(preds == labels.data)
                progress.update(inputs.size(0))
            progress.close()

            if phase == "train" and scheduler is not None:
                scheduler.step()

            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]
            epoch_losses[phase] = float(epoch_loss)
            epoch_accs[phase] = float(epoch_acc)
            print(f"{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}")

            if phase == "val" and epoch_acc > best_val_acc:
                best_val_acc = float(epoch_acc)
                best_epoch = epoch
                improved_this_epoch = True
                best_model_wts = copy.deepcopy(model.state_dict())
                torch.save({
                    "arch": arch, "model_name": model_name, "state_dict": model.state_dict(),
                    "class_names": CLASS_NAMES, "major_names": MAJOR_NAMES,
                    "class_to_major": CLASS_TO_MAJOR, "img_size": IMG_SIZE,
                    "mean": IMAGENET_MEAN, "std": IMAGENET_STD,
                    "epoch": epoch, "val_acc": float(epoch_acc), "val_loss": float(epoch_loss),
                    "config": {"seed": SEED, "batch_size": BATCH_SIZE, "epochs": epochs, "pretrained": PRETRAINED, **(extra_config or {})},
                }, checkpoint_path)

        if improved_this_epoch:
            epochs_without_improvement = 0
        else:
            epochs_without_improvement += 1

        current_lr = optimizer.param_groups[0]["lr"]
        record = {
            "epoch": epoch, "train_loss": epoch_losses["train"], "train_acc": epoch_accs["train"],
            "val_loss": epoch_losses["val"], "val_acc": epoch_accs["val"],
            "lr": float(current_lr), "time_sec": float(time.time() - epoch_start), "best_val_acc": best_val_acc,
        }
        history.append(record)
        pd.DataFrame(history).to_csv(history_path, index=False)
        print(f"[{model_name}] epoch {epoch:02d}/{epochs} | train acc {record['train_acc']:.4f} | val acc {record['val_acc']:.4f} | lr {current_lr:.2e}")
        print()

        if early_stopping_patience is not None and epoch >= early_stop_min_epochs and epochs_without_improvement >= early_stopping_patience:
            stopped_epoch = epoch
            print(f"Early stopping at epoch {epoch} (no improvement for {early_stopping_patience} epochs).")
            break

    time_elapsed = time.time() - since
    print(f"Training complete in {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s | best val {best_val_acc:.4f} @ {best_epoch}")
    model.load_state_dict(best_model_wts)
    if full_model_path is not None:
        torch.save(model, full_model_path)
    print(f"Checkpoint: {checkpoint_path}")
    return history, best_val_acc, best_epoch


def load_checkpoint(path, device):
    try:
        return torch.load(path, map_location=device, weights_only=True)
    except TypeError:
        return torch.load(path, map_location=device)
    except Exception:
        return torch.load(path, map_location=device, weights_only=False)
```

## 6. Dua varian training

- **B · cosine** (`mobilenetv2_cosine`): cosine annealing, label smoothing 0.1, weight decay 1e-4, augmentasi standar
- **Cosine + Augment** (`mobilenetv2_cosine_aug`): cosine annealing, label smoothing 0.1, weight decay 1e-4, augmentasi kuat

Split, seed, arsitektur, epoch, dan early stopping identik. File artefak memakai prefix di atas (tidak menimpa baseline `mobilenetv2_*`).

```python
VARIANTS = {
    "mobilenetv2_cosine": {
        "label": "B · cosine",
        "transform": train_transform,
        "label_smoothing": 0.1,
        "weight_decay": 1e-4,
        "scheduler": "cosine",
        "class_weight": False,
        "augmentation": "randomresizedcrop+flip",
    },
    "mobilenetv2_cosine_aug": {
        "label": "Cosine + Augment",
        "transform": aug_train_transform,
        "label_smoothing": 0.1,
        "weight_decay": 1e-4,
        "scheduler": "cosine",
        "class_weight": False,
        "augmentation": "rrc+flip+trivialaugment+colorjitter+randomerasing",
    },
}


def build_criterion(spec):
    if spec.get("class_weight"):
        counts = train_df["class_name"].value_counts().reindex(CLASS_NAMES).to_numpy()
        weights = torch.tensor(counts.sum() / (NUM_CLASSES * counts), dtype=torch.float32, device=DEVICE)
        return nn.CrossEntropyLoss(weight=weights, label_smoothing=spec.get("label_smoothing", 0.0))
    return nn.CrossEntropyLoss(label_smoothing=spec.get("label_smoothing", 0.0))


def build_scheduler(optimizer, spec, epochs):
    if spec.get("scheduler") == "cosine":
        return lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)
    return lr_scheduler.StepLR(optimizer, step_size=STEP_SIZE, gamma=GAMMA)


for name, spec in VARIANTS.items():
    print(f"{name:26s} | {spec['label']:18s} | scheduler={spec['scheduler']} ls={spec['label_smoothing']} wd={spec['weight_decay']} aug={spec['augmentation']}")
```

```python
runs = {}

for name, spec in VARIANTS.items():
    print("=" * 78)
    print(f"Training {name} ({spec['label']})")
    print("=" * 78)

    variant_train_loader = make_loader(train_df, training=True, batch_size=BATCH_SIZE, transform=spec["transform"])
    variant_model = build_mobilenet_v2(NUM_CLASSES, pretrained=PRETRAINED).to(DEVICE)
    variant_criterion = build_criterion(spec)
    variant_optimizer = optim.SGD(
        variant_model.parameters(), lr=LEARNING_RATE, momentum=MOMENTUM,
        weight_decay=spec.get("weight_decay", 0.0),
    )
    variant_scheduler = build_scheduler(variant_optimizer, spec, NUM_EPOCHS)

    checkpoint_path = MODEL_DIR / f"{name}_trash30_best.pt"
    full_model_path = MODEL_DIR / f"{name}_trash30_full.pt"
    history_path = RESULT_DIR / f"{name}_history.csv"

    variant_history, variant_best_acc, variant_best_epoch = train_model(
        model=variant_model, train_loader=variant_train_loader, val_loader=val_loader,
        criterion=variant_criterion, optimizer=variant_optimizer, scheduler=variant_scheduler,
        epochs=NUM_EPOCHS, device=DEVICE,
        checkpoint_path=checkpoint_path, full_model_path=full_model_path, history_path=history_path,
        model_name=name, arch="mobilenet_v2",
        extra_config={
            "variant": spec["label"], "learning_rate": LEARNING_RATE, "momentum": MOMENTUM,
            "weight_decay": spec.get("weight_decay", 0.0), "label_smoothing": spec.get("label_smoothing", 0.0),
            "scheduler": spec.get("scheduler", "step"), "class_weight": spec.get("class_weight", False),
            "augmentation": spec.get("augmentation", ""),
        },
        early_stopping_patience=EARLY_STOP_PATIENCE, early_stop_min_epochs=EARLY_STOP_MIN_EPOCHS,
    )

    runs[name] = {
        "label": spec["label"], "spec": spec, "model": variant_model,
        "history": variant_history, "best_val_acc": variant_best_acc, "best_epoch": variant_best_epoch,
        "checkpoint": checkpoint_path, "full_model": full_model_path, "history_csv": history_path,
    }
    print(f"{name}: best val acc {variant_best_acc:.4f} @ epoch {variant_best_epoch}")

print("\nSelesai:", ", ".join(runs))
```

## 8. Evaluasi test

```python
def jsonable(obj):
    if isinstance(obj, dict): return {str(k): jsonable(v) for k, v in obj.items()}
    if isinstance(obj, (list, tuple)): return [jsonable(v) for v in obj]
    if isinstance(obj, np.ndarray): return obj.tolist()
    if isinstance(obj, np.integer): return int(obj)
    if isinstance(obj, np.floating): return float(obj)
    if isinstance(obj, Path): return str(obj)
    return obj


def aggregate_major_probs(probs: np.ndarray, class_names):
    major_probs = np.zeros((probs.shape[0], len(MAJOR_NAMES)), dtype=np.float64)
    for class_idx, class_name in enumerate(class_names):
        major_probs[:, MAJOR_TO_IDX[CLASS_TO_MAJOR[class_name]]] += probs[:, class_idx]
    return major_probs


@torch.no_grad()
def predict_loader(model, loader, criterion, device):
    model.eval()
    all_probs, all_labels = [], []
    total_loss = 0.0
    total_samples = 0
    for images, labels in tqdm(loader, desc="Evaluating", leave=False):
        images = images.to(device, non_blocking=True)
        labels = labels.to(device, non_blocking=True)
        outputs = model(images)
        loss = criterion(outputs, labels)
        probs = torch.softmax(outputs, dim=1)
        all_probs.append(probs.cpu().numpy())
        all_labels.append(labels.cpu().numpy())
        total_loss += loss.item() * labels.size(0)
        total_samples += labels.size(0)
    probs = np.concatenate(all_probs, axis=0)
    labels = np.concatenate(all_labels, axis=0)
    return probs, labels, total_loss / total_samples


def plot_confusion_matrix(cm, labels, title, output_path, normalize=True):
    cm = np.asarray(cm, dtype=np.float64)
    if normalize:
        row_sums = cm.sum(axis=1, keepdims=True)
        plot_cm = np.divide(cm, row_sums, out=np.zeros_like(cm), where=row_sums != 0)
        fmt = ".2f"
    else:
        plot_cm = cm
        fmt = "d"
    annot = len(labels) <= 12
    fig, ax = plt.subplots(figsize=(max(7, len(labels) * 0.48), max(6, len(labels) * 0.42)))
    sns.heatmap(plot_cm, annot=annot, fmt=fmt, cmap="Blues", xticklabels=labels, yticklabels=labels,
                ax=ax, cbar=True, linewidths=0.2 if annot else 0, annot_kws={"fontsize": 9} if annot else None)
    ax.set_xlabel("predicted"); ax.set_ylabel("true"); ax.set_title(title)
    ax.tick_params(axis="x", rotation=90, labelsize=7); ax.tick_params(axis="y", rotation=0, labelsize=7)
    fig.tight_layout()
    fig.savefig(output_path, bbox_inches="tight")
    plt.show()
    plt.close(fig)


def evaluate_and_save(model, loader, criterion, device, frame, model_key):
    probs, labels, loss = predict_loader(model, loader, criterion, device)
    preds = probs.argmax(axis=1)
    report_dict = classification_report(labels, preds, labels=np.arange(NUM_CLASSES), target_names=CLASS_NAMES, output_dict=True, zero_division=0)
    cm = confusion_matrix(labels, preds, labels=np.arange(NUM_CLASSES))
    per_class_acc = np.divide(cm.diagonal(), cm.sum(axis=1), out=np.zeros(NUM_CLASSES, dtype=np.float64), where=cm.sum(axis=1) != 0)

    major_probs = aggregate_major_probs(probs, CLASS_NAMES)
    major_true = frame["class_name"].map(lambda c: MAJOR_TO_IDX[CLASS_TO_MAJOR[c]]).to_numpy()
    major_pred = np.array([MAJOR_TO_IDX[CLASS_TO_MAJOR[CLASS_NAMES[int(i)]]] for i in preds])
    major_prob_pred = major_probs.argmax(axis=1)
    major_report = classification_report(major_true, major_pred, labels=np.arange(len(MAJOR_NAMES)), target_names=MAJOR_NAMES, output_dict=True, zero_division=0)
    major_cm = confusion_matrix(major_true, major_pred, labels=np.arange(len(MAJOR_NAMES)))

    prediction_frame = frame[["rel_path", "class_name", "source_group"]].copy()
    prediction_frame["true_label"] = labels
    prediction_frame["pred_label"] = preds
    prediction_frame["pred_class"] = [CLASS_NAMES[i] for i in preds]
    prediction_frame["confidence"] = probs.max(axis=1)
    prediction_frame["correct"] = (labels == preds).astype(int)
    for class_idx, class_name in enumerate(CLASS_NAMES):
        prediction_frame[f"prob_{class_name}"] = probs[:, class_idx]
    prediction_frame.to_csv(RESULT_DIR / f"{model_key}_predictions.csv", index=False)

    metrics = {
        "model": model_key, "split": "test", "num_test": int(len(labels)),
        "accuracy": float(accuracy_score(labels, preds)), "loss": float(loss),
        "top3_accuracy": float(top_k_accuracy_score(labels, probs, k=min(3, NUM_CLASSES), labels=np.arange(NUM_CLASSES))),
        "macro_f1": float(report_dict["macro avg"]["f1-score"]), "weighted_f1": float(report_dict["weighted avg"]["f1-score"]),
        "per_class_accuracy": {name: float(per_class_acc[i]) for i, name in enumerate(CLASS_NAMES)},
        "group_accuracy": {str(group): float(sub["correct"].mean()) for group, sub in prediction_frame.groupby("source_group")},
        "classification_report": report_dict, "confusion_matrix": cm.tolist(),
        "major_accuracy": float(accuracy_score(major_true, major_pred)),
        "major_prob_accuracy": float(accuracy_score(major_true, major_prob_pred)),
        "major_report": major_report, "major_confusion_matrix": major_cm.tolist(),
        "major_class_names": MAJOR_NAMES, "class_names": CLASS_NAMES,
    }
    with open(RESULT_DIR / f"{model_key}_metrics.json", "w") as handle:
        json.dump(jsonable(metrics), handle, indent=2)
    pd.DataFrame(report_dict).T.to_csv(RESULT_DIR / f"{model_key}_classification_report.csv")
    pd.DataFrame(major_report).T.to_csv(RESULT_DIR / f"{model_key}_major_report.csv")
    pd.DataFrame(per_class_acc, index=CLASS_NAMES, columns=["accuracy"]).to_csv(RESULT_DIR / f"{model_key}_per_class_accuracy.csv")

    print(f"\n{model_key} — test acc {metrics['accuracy']*100:.2f}% | macro F1 {metrics['macro_f1']:.4f} | top-3 {metrics['top3_accuracy']*100:.2f}% | 3-class {metrics['major_accuracy']*100:.2f}%")
    for group, value in metrics["group_accuracy"].items():
        print(f"  {group}: {value*100:.2f}%")
    return metrics, probs, labels, preds
```

```python
eval_criterion = nn.CrossEntropyLoss()
evaluations = {}

for name, run in runs.items():
    eval_model = build_mobilenet_v2(NUM_CLASSES, pretrained=False).to(DEVICE)
    ckpt = load_checkpoint(run["checkpoint"], DEVICE)
    eval_model.load_state_dict(ckpt["state_dict"])
    eval_model.eval()
    print(f"\n{name}: checkpoint epoch {ckpt['epoch']} (val acc {ckpt['val_acc']:.4f})")
    metrics, test_probs, test_labels, test_preds = evaluate_and_save(
        eval_model, test_loader, eval_criterion, DEVICE, test_df, model_key=name
    )
    evaluations[name] = {"metrics": metrics, "model": eval_model, "probabilities": test_probs, "labels": test_labels, "predictions": test_preds}
    display(pd.DataFrame(metrics["major_report"]).T.round(3))
```

## 10. Analisis error

```python
def build_prediction_frame(frame, probs, labels, preds):
    result = frame[["rel_path", "class_name", "source_group"]].copy()
    result["true_class"] = [CLASS_NAMES[int(i)] for i in labels]
    result["pred_class"] = [CLASS_NAMES[int(i)] for i in preds]
    result["confidence"] = probs.max(axis=1)
    result["correct"] = result["true_class"].eq(result["pred_class"])
    return result

best_name = max(evaluations, key=lambda k: evaluations[k]["metrics"]["accuracy"])
for name, evaluation in evaluations.items():
    predictions = build_prediction_frame(test_df, evaluation["probabilities"], evaluation["labels"], evaluation["predictions"])
    errors = predictions[~predictions["correct"]].sort_values("confidence", ascending=False).reset_index(drop=True)
    print(f"{runs[name]['label']}: {len(errors):,} salah dari {len(predictions):,}")
    errors.to_csv(RESULT_DIR / f"{name}_misclassified.csv", index=False)
    if not errors.empty and name == best_name:
        examples = errors.head(6)
        ncols = min(3, len(examples))
        nrows = math.ceil(len(examples) / ncols)
        fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 4.2 * nrows))
        axes = np.atleast_1d(axes).ravel()
        for ax, (_, row) in zip(axes, examples.iterrows()):
            with Image.open(DATA_ROOT / row["rel_path"]) as img:
                ax.imshow(img.convert("RGB"))
            ax.set_title(f"True: {row['true_class']}\nPred: {row['pred_class']} ({row['confidence']*100:.1f}%)", fontsize=8, color="darkred")
            ax.axis("off")
        for ax in axes[len(examples):]:
            ax.axis("off")
        fig.suptitle(f"{runs[name]['label']} misclassified test examples")
        fig.tight_layout()
        fig.savefig(FIGURE_DIR / f"{name}_misclassified_examples.png", bbox_inches="tight")
        plt.show()
        plt.close(fig)
```
